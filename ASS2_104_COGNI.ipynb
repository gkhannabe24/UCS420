```python
# Q1. List Operations

roll_number = 1024170104
digits = [int(i) for i in str(roll_number)]

L = [digit * 10 for digit in digits]
print("Initial List L:", L)

L.append(90)
print("After append:", L)

L.insert(2, 50)
print("After insert:", L)

L.remove(50)
print("After remove:", L)

L.pop()
print("After pop:", L)

L.sort()
print("Ascending order:", L)

L.sort(reverse=True)
print("Descending order:", L)

print("First three elements:", L[:3])
print("Last three elements:", L[-3:])

average = sum(L)/len(L)
greater_than_average = [x for x in L if x > average]
print("Elements greater than average:", greater_than_average)
```

```python
# Q2. Tuple Operations

scores = tuple(L[:8])
print("Scores tuple:", scores)

print("Highest score:", max(scores))
print("Index of highest score:", scores.index(max(scores)))

print("Lowest score:", min(scores))
print("Frequency of lowest score:", scores.count(min(scores)))

reversed_scores = list(reversed(scores))
# Tuple is immutable, so it cannot be reversed in place.
print("Reversed tuple as list:", reversed_scores)

search_score = int(input("Enter a score to search: "))
if search_score in scores:
    print("First occurrence index:", scores.index(search_score))
else:
    print("Score not present")

# Trying to modify a tuple element gives TypeError because tuples are immutable.
try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)
    print("Tuples cannot be changed after creation, unlike lists.")

first, second, *remaining = scores
print(first, second, remaining)

```

```python
# Q3. Random Number Operations

import random
import math
from collections import Counter

random.seed(12345678)

numbers = [random.randint(100,900) for _ in range(100)]

odd_numbers = [x for x in numbers if x % 2 != 0]
even_numbers = [x for x in numbers if x % 2 == 0]

print("Odd numbers count:", len(odd_numbers))
print("Odd numbers:", odd_numbers)

print("Even numbers count:", len(even_numbers))
print("Even numbers:", even_numbers)

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(math.sqrt(n))+1):
        if n % i == 0:
            return False
    return True

prime_numbers = [x for x in numbers if is_prime(x)]

print("Prime count:", len(prime_numbers))
print("Prime numbers:", prime_numbers)

frequency = Counter(numbers)
most_common_number = frequency.most_common(1)[0]

print("Most frequent number:", most_common_number[0])
print("Frequency:", most_common_number[1])
```

```python
# Q4. Set Operations

digits = [int(i) for i in str(12345678)]

A = {digit*7 for digit in digits}
B = {digit*9 for digit in digits}

print("Set A:", A)
print("Set B:", B)

print("Union:", A.union(B))
print("Intersection:", A.intersection(B))

print("A - B:", A.difference(B))
print("B - A:", B.difference(A))

print("Symmetric Difference:", A.symmetric_difference(B))
# difference() gives one-sided difference, symmetric_difference() gives all non-common values.

print("A subset of B:", A.issubset(B))
print("B superset of A:", B.issuperset(A))

x = int(input("Enter value to remove from A: "))
A.discard(x)
# discard() is safer because it does not raise an error if the value is absent.
print("Updated A:", A)
```

```python
# Q5. Dictionary Operations

my_dict = {
    "name": "Gorv",
    "roll_no": "1024170104",
    "branch": "CSE",
    "age": 20,
    "city": "Patiala"
}

my_dict["location"] = my_dict.pop("city")

my_dict["cgpa"] = 9.0

my_dict["age"] += 1

print("Updated dictionary:", my_dict)

dict1 = my_dict.copy()
removed = dict1.pop("branch")
print("After pop:", dict1)
# pop() returns the removed value, while del only deletes the key.

dict2 = my_dict.copy()
del dict2["branch"]
print("After del:", dict2)

for key, value in my_dict.items():
    print(key, "→", value)

if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email key does not exist")

friend_dict = {
    "name": "sam",
    "roll_no": "1024170099",
    "branch": "CSE",
    "age": 19,
    "city": "patiala"
}

merged = {**my_dict, **friend_dict}
# Values from the second dictionary overwrite matching keys.
print("Merged dictionary:", merged)

string_values = {k:v for k,v in my_dict.items() if isinstance(v,str)}
print("Only string values:", string_values)
```
